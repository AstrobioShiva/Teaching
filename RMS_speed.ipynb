```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(M, T, v):
    R = 8.314  # J/(mol*K)
    return 4*np.pi*(M/(2*np.pi*R*T))**(3/2)*v**2*np.exp(-M*v**2/(2*R*T))
```

```python
fig = plt.figure(figsize =(8, 6)) 
fig.set_dpi(100)
v = np.linspace(0, 150, 20)
m = 32*10**-3  # kg/mol for O2
T_300 = 300  # K
T_80 = 80  # K
f_300_values = f(m, T_300, v)*10**3
f_80_values = f(m, T_80, v)*10**3
plt.plot(v, f_300_values, label='300 K', lw = 3)
plt.plot(v, f_80_values, label='80 K', lw = 3)
plt.xlabel('Velocity (m/s)', fontsize=15)
plt.ylabel('Probability Density', fontsize=15)
plt.title('Maxwell-Boltzmann Distribution for O2')
plt.legend(fontsize=15)
plt.grid()
plt.show()
```
