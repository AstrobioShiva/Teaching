```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def integral(f,a,b,dx):
    p = np.arange(a,b,dx)
    q = f(p)
    return np.sum(q*dx)

func = lambda y: y**2                                   #Choosing the function

l_limit = 0                                             #lower limit
u_limit = 10                                            #upper limit
step_size = 0.01                                        #size of rectangle

estimate = integral(func, l_limit, u_limit, step_size)
actual = (u_limit**3 - l_limit**3) / 3                  #actual integral of y^2 from 0 to 10
print("Estimated integral:", estimate)
print("Actual integral:", actual)


    
    
```

```python
plt.figure(figsize=(10,5), dpi=200)
x = np.linspace(l_limit, u_limit, 100)
plt.plot(x, func(x), label='f(y) = y^2', color='blue')
plt.title('Integral of f(y) = y^2 from {} to {}'.format(l_limit, u_limit))
plt.xlabel('y')
plt.ylabel('f(y)')
plt.legend()
plt.grid()
plt.show()
```
