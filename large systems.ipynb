```python
import numpy as np
import matplotlib.pyplot as plt
import math 
```

```python
x = np.linspace(0, 5, 100)

y_exp = np.log(x)

plt.figure(figsize=(5,5), dpi=100)
plt.plot(x,y_exp, label = '$y = ln(x)$')
plt.title('Graph of Natural Logarithmic Function')
plt.xlabel('x')
plt.ylabel('ln(x)')
plt.legend()
plt.grid()
```

```python
x = np.linspace(0, 0.1, 100)

y_exp = np.log(1+x)
y_line = x

plt.figure(figsize=(5,5), dpi=100)
plt.plot(x,y_exp, label = '$y = ln(x)$')
plt.plot(x,y_line, label = '$y = x$')
plt.title('Graph of Natural Logarithmic Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
```

```python
N = 100
x = np.linspace(1, N, N*100) #starting from 1 to avoid log(0) error
y1 = [math.factorial(int(i)) for i in x] #factorial function
y2 = [(i**i)*(np.exp(-i))*(np.sqrt(2*np.pi*i)) for i in x] #Sterling's approximation

y3 = [math.log(math.factorial(int(i))) for i in x] #log factorial function
y4 = [i*np.log(i) - i for i in x] #log Sterling's approximation
```

**Plotting the Functions**

```python
plt.figure(figsize=(20,5), dpi=100)
plt.subplot(1, 2, 1)
plt.plot(x,y1, label = '$factorial$')
plt.plot(x,y2, label = '$Stirling$')
plt.title('Graph to verify Stirling\'s approximation for factorial')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()


plt.subplot(1, 2, 2)
plt.plot(x,y3, label = '$log(factorial)$')
plt.plot(x,y4, label = '$log(Stirling)$')
plt.title('Graph to verify Stirling\'s approximation for log(factorial)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
plt.show()
```
