```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5, 100)

y_exp = np.log(x)

plt.figure(figsize=(5,5), dpi=100)
plt.plot(x,y_exp, label = '$y = ln(x)$')
plt.title('Graph of Natural Logarithmic Function')
plt.xlabel('x')
plt.ylabel('ln(x)')
plt.legend()
plt.grid()
```

```python
x = np.linspace(0, 0.1, 100)

y_exp = np.log(1+x)
y_line = x

plt.figure(figsize=(5,5), dpi=100)
plt.plot(x,y_exp, label = '$y = ln(x)$')
plt.plot(x,y_line, label = '$y = x$')
plt.title('Graph of Natural Logarithmic Function')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
```
