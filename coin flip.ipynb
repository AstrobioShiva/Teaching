```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
def omega(N,n):
   
    return math.factorial(N) / (math.factorial(n) * math.factorial(N - n))

N = 50
n = 10
multiplicity = omega(N,n)

print("Multiplicity of the system:", multiplicity)
```

```python
N = 20 # number of coins

n_values = np.linspace(0, N, N+1) # generate evenly spaced values of n from 0 to N
multiplicity = [omega(N, n) for n in n_values] # calculate multiplicity for each n

plt.figure(figsize=(10,5), dpi=200)
x = n_values
plt.plot(x, multiplicity, color='blue', marker = 'o')
plt.title('Multiplicity vs Number of Heads in Coin Flips for {} coins'.format(N))
plt.xlabel('Number of Heads/Tails')
plt.ylabel('Multiplicity')
# plt.legend()
plt.show()
```
